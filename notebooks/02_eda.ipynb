# Week 5 — Exploratory Data Analysis

> **DEMO / SAMPLE DATA NOTICE:** This notebook uses a synthetic demonstration dataset created for the Week 5 reporting structure. It must not be presented as actual internship findings or real Week 1–4 results.


```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/processed/cleaned_dataset.csv")
display(df.describe(include="all").T)
```

```python
print("Sales by region:")
display(df.groupby("region")["sales"].agg(["sum","mean","count"]).sort_values("sum", ascending=False))

print("Sales by category:")
display(df.groupby("category")["sales"].agg(["sum","mean","count"]).sort_values("sum", ascending=False))
```

```python
print("Correlation matrix:")
display(df.select_dtypes(include=np.number).corr().round(2))
```
