# Week 5 — Data Cleaning

> **DEMO / SAMPLE DATA NOTICE:** This notebook uses a synthetic demonstration dataset created for the Week 5 reporting structure. It must not be presented as actual internship findings or real Week 1–4 results.


```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/dataset.csv")
print(df.shape)
display(df.head())
```

```python
# Inspect missing values and duplicates
print(df.isna().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Demonstration cleaning rules
df["category"] = df["category"].fillna(df["category"].mode()[0])
df["unit_price"] = df["unit_price"].fillna(df["unit_price"].median())
df["profit"] = df["profit"].fillna(df["profit"].median())
df = df.drop_duplicates()

df.to_csv("../data/processed/cleaned_dataset.csv", index=False)
print("Cleaned shape:", df.shape)
```
